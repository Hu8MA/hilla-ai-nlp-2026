```python
text="The NLP workshop starts on 10/15/2025 in Hall 3. Dr. Ali and Dr. Sara will teach it. The fee is $25 or $15 for students. Call 07701234567 or 0780-123-4567 to register. Last year the the workshop had 120 students; this year we expect 150. Is the room ready? We hope so! The other rooms are closed, but there is a colour printer and a color screen."
```

```python
import re
```

```python
# Q1. Find all the numbers in the paragraph

print(re.findall(r"\d+", text))
```

```python
# Q2. Find all words made of a capital letter followed by lowercase letters


print(re.findall(r"\b[A-Z][a-z]+\b", text))
```

```python
# Q3. Find the word "the" or "The" as a whole word only


print(re.findall(r"\b[tT]he\b", text))
```

```python
# Q4. Find both spellings "color" and "colour" with one pattern


print(re.findall(r"colou?r", text))
```

```python
# Q5. Find all prices (a dollar sign followed by digits)


print(re.findall(r"\$\d+", text))

## note: the original paragraph text was missing the '$' symbol (25 or 15),so i added it.
```

```python
# Q6. Find the two phone numbers


print(re.findall(r"07\d{2}[-]?\d{3}[-]?\d{4}", text))
```

```python
# Q7. Find the titles with names, like "Dr. Ali"


print(re.findall(r"Dr\.\s[A-Z][a-z]+", text))
```

```python
# Q8. Find the word that is repeated twice in a row


print(re.findall(r"\b([A-Za-z]+)\s+\1\b", text))
```

```python
# Q9. Change the date from mm/dd/yyyy to dd-mm-yyyy


print(re.sub(r"(\d{2})/(\d{2})/(\d{4})", r"\2-\1-\3", text))
```

```python
# Q10. Tokenize the paragraph into words and punctuation marks


tokens = re.findall(r"\w+|[^\w\s]", text)
print(tokens)
print("Total tokens:", len(tokens))
```
