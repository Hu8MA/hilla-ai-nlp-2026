Q1\

Find one sentence (English or Arabic) that has two meanings. Write both meanings and explain in one or two lines why a computer would find it hard to choose the right one.


answer:

"I saw Professor Hussein teaching students with glasses"

Meaning 1:
I saw Professor Hussein (who wears glasses) teaching the students.

Meaning 2:
I saw Professor Hussein teaching students who were wearing glasses.


the computer struggles here because the phrase "with glasses" can grammatically translate to the verb subject (Prof. Hussein) or the object (students), the ai cannot know who is actually wearing the glasses.
