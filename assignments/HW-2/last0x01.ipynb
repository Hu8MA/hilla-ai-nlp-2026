```python
import re
```

### Regex Exercises

- **Q1.** Find all the numbers in the paragraph (one or more digits together).

```python
text = """The NLP workshop starts on 10/15/2025 in Hall 3. Dr. Ali and Dr. Sara will teach it.
The fee is $25, or $15 for students. Call 07701234567 or 0780-123-4567 to register.
Last year the the workshop had 120 students; this year we expect 150. Is the room ready?
We hope so! The other rooms are closed, but there is a colour printer and a color screen."
"""

print(re.findall(r'\d+',text))
```

- **Q2.** Find all words made of a capital letter followed by lowercase letters (like "Hall").

```python
print(re.findall(r'[A-Z][a-z]+',text))
```

- **Q3.** Find the word "the" or "The" as a whole word only (not inside "other" or "there").

```python
print(re.findall(r'\b[tT]he\b',text))
```

- **Q4.** Find both spellings "color" and "colour" with one pattern.

```python
print(re.findall(r'colou?r',text))
print(re.findall(r'colour|color',text))
```

- **Q5.** Find all prices (a dollar sign followed by digits).

```python
print(re.findall(r'\$\d+',text))
```

- **Q6.** Find the two phone numbers (07, then 9 digits, with or without dashes).

```python
print(re.findall(r'07\d{9}',text))
print(re.findall(r'07\d{2}-\d{3}-\d{4}',text))
```

- **Q7.** Find the titles with names, like "Dr. Ali".

```python
print(re.findall(r'Dr\.\s[A-Z][a-z]+',text))
```

- **Q8.** Find the word that is repeated twice in a row.

```python
print(re.findall(r"\b([A-Za-z]+)\s+\1\b", text))
```

- **Q9.** Change the date from mm/dd/yyyy to dd-mm-yyyy.

```python
old_date = "10/15/2025"
print(old_date)

new_date = re.sub(r'(\d{2})/(\d{2})/(\d{4})', r'\2-\1-\3', old_date)
print(new_date)
```

- **Q10.** Tokenize the paragraph: split it into words and punctuation marks, then print how many tokens there are.

```python
tokens = re.findall(r"\w+|[^\w\s]", text)
print(tokens, "\n\nNumber of tokens:", len(tokens))
```
